```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("titanic.csv")

df.head()
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="Survived")

plt.title("Titanic Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="Sex", hue="Survived")

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="Pclass", hue="Survived")

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x="Age", bins=20, kde=True)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x="Fare", bins=20, kde=True)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

```python
gender_survival = df.groupby("Sex")["Survived"].mean() * 100

print(gender_survival)
```

```python
plt.figure(figsize=(7,5))

gender_survival.plot(kind="bar")

plt.title("Survival Percentage by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Percentage (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(7,5))

gender_survival.plot(kind="bar")

plt.title("Survival Percentage by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Percentage (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
class_survival = df.groupby("Pclass")["Survived"].mean() * 100

print(class_survival)
```

```python
plt.figure(figsize=(7,5))

class_survival.plot(kind="bar")

plt.title("Survival Percentage by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Percentage (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
correlation = df[
    ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]
].corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Titanic Dataset Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(data=df, y="Fare")

plt.title("Fare Distribution and Outliers")
plt.ylabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(data=df, y="Age")

plt.title("Age Distribution and Outliers")
plt.ylabel("Age")

plt.show()
```
